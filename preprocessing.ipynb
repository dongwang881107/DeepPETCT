```python
import pydicom
import glob
import os
import numpy as np
import shutil
import random
```

```python
data_path = r'G:\1012010'
dest_path = r'G:\sandwich'
```

```python
# For old 10s-60s data
def get_files(folder_path, dest_path):
    files = os.listdir(folder_path)
    for file in files:
        file_path = os.path.join(folder_path,file)
        if os.path.isdir(file_path):
            get_files(file_path, dest_path)
        else:
            if file == 'I10':
                try:
                    dcm = pydicom.dcmread(file_path)
                    patient_path = dest_path+'\\'+str(dcm.PatientID)
                    if not os.path.exists(patient_path):
                        os.mkdir(patient_path)
                    try:
                        dcm_series = str(dcm[0x7053,0x1079].value)
                        if dcm_series == '10':
                            dcm_series = 'short_10s'
                        elif dcm_series == '60':
                            dcm_series = 'long_60s'
                        else:
                            dcm_series = ''
                    except:
                        dcm_series = 'ct'
                    series_path = patient_path+'\\'+dcm_series
                    try:
                        shutil.copytree('\\'.join(file_path.split('\\')[:-1]),series_path)
                    except:
                        print('Already Exist!')
                except:
                    continue
```

```python
data_path = r'G:\1012010'
dest_path = r'G:\sandwich'
```

```python
# For new sandwich data
def get_files(folder_path, dest_path):
    files = os.listdir(folder_path)
    for file in files:
        file_path = os.path.join(folder_path,file)
        if os.path.isdir(file_path):
            get_files(file_path, dest_path)
        else:
            if file == 'I10':
                file_size = os.path.getsize(file_path)
                if file_size < 1532114:
                    try:
                        dcm = pydicom.dcmread(file_path)
                        patient_path = dest_path+'\\'+str(dcm.PatientID)
                        if not os.path.exists(patient_path):
                            os.mkdir(patient_path)
                        try:
                            dcm_series = str(dcm[0x7053,0x1079].value)
                            if dcm_series == '10':
                                times_tag = str(dcm.SeriesTime)
                                dcm_series = 'short_10s_'+times_tag
                            elif dcm_series == '120':
                                dcm_series = 'long_120s'
                            else:
                                dcm_series = ''
                        except:
                            dcm_series = 'ct'
                        series_path = patient_path+'\\'+dcm_series
                        try:
                            shutil.copytree('\\'.join(file_path.split('\\')[:-1]),series_path)
                        except:
                            print('Already Exist!')
                    except:
                        continue
                        
# run after get_files
def clean_files(folder_path):
    case_paths = glob.glob(folder_path+'/P*')
    # delete useless files
    for case_path in case_paths:
        series_paths = glob.glob(case_path+'/*')
        for series_path in series_paths:
            files_to_delete = glob.glob(series_path+'/.*')
            for file in files_to_delete:
                os.remove(file)           
    # rename 10s folders
    for case_path in case_paths:
        series_folders = os.listdir(case_path)
        short_folders = []
        short_folders_int = []
        renamed_folders = [None]*2
        for series_folder in series_folders:
            if ('short' in series_folder) & ('p' not in series_folder):
                short_folders.append(series_folder)
                series_folder = int(series_folder.split('_')[-1])
                short_folders_int.append(series_folder)
        if short_folders_int:
            if short_folders_int[0] > short_folders_int[1]:
                renamed_folders[0] = 'short_10s_post'
                renamed_folders[1] = 'short_10s_pre'
            else:
                renamed_folders[0] = 'short_10s_pre'
                renamed_folders[1] = 'short_10s_post'
            for i in range(2):
                os.rename(case_path+'\\'+short_folders[i],case_path+'\\'+renamed_folders[i])
        
```
